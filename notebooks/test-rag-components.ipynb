# Tiempos de respuesta

```python
import time
import statistics as stats
from huggingface_hub import InferenceClient

# Import from your package (adjust the module path)
from app.rag import rag_answer
from app.retrieval import search

HF_API_KEY = settings.hf_api_key
client = InferenceClient(api_key=HF_API_KEY, provider="novita")  # adjust provider if needed

def timed_once(query: str, top_k: int = 5, model: str = "meta-llama/Llama-3.1-8B-Instruct"):
    chat_history = []  # or include a small history if you want

    t0 = time.perf_counter()
    chunks = search(query, top_k=top_k)
    t1 = time.perf_counter()

    # Call rag_answer (will retrieve again). If you want *separate* timing, don't use rag_answer here.
    # So we do the "RAG call" ourselves to avoid double retrieval.
    from app.rag import SYSTEM_PROMPT, build_context_block

    context = build_context_block(chunks, verbose=False)
    messages = [{"role": "system", "content": SYSTEM_PROMPT}]
    messages.extend(chat_history[-10:])
    messages.append({"role": "user", "content": f"CONTEXT:\n{context}\n\nQUESTION:\n{query}"})

    t2 = time.perf_counter()
    completion = client.chat.completions.create(
        model=model,
        messages=messages,
        max_tokens=500,
    )
    t3 = time.perf_counter()

    answer = completion.choices[0].message["content"]

    return {
        "retrieve_s": t1 - t0,
        "llm_s": t3 - t2,
        "total_s": t3 - t0,
        "n_chunks": len(chunks),
        "answer": answer,
    }
```

```python
# Single run
res = timed_once(user_message_1, top_k=5)
res
```
